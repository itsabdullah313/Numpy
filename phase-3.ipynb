## Advance Operations with Business Examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure : [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 15000, 18000, 22000, 25000], # Paradise Biryani
    [2, 12000, 14000, 16000, 19000], # Beijing Bites
    [3, 20000, 23000, 26000, 30000], # Pizza Hub
    [4, 18000, 21000, 24000, 27000], # Burger Point
    [5, 16000, 18500, 20500, 23000]  # Chai Point
])
print("Data Shape ", sales_data.shape)
print("First 3 data ", sales_data[:3])
```

```python
# Total sales per year
print(np.sum(sales_data[:,1:], axis=0))

# Minimum sales per restaurant
print("Min sale per restaurant ", np.min(sales_data[:,1:], axis=1))

# Maximum sale per year
print("Max Sale per year ", np.max(sales_data[:,1:], axis=0))

# Average sales per restaurant
print("Average sale per restaurant ", np.mean(sales_data[:,1:],axis=1))

# Cumulative sales
cumsum = np.cumsum(sales_data[:,1:], axis=1)
print("Cumulative Sales ", cumsum)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum ,axis=0))
plt.title("Average cumulative sales across all restaurants")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])
print("Addition ", vector1 + vector2)
print("Multiplication ", vector1 * vector2)
print("Dot Product ", np.dot(vector1,vector2))
angle = np.arccos(np.dot(vector1,vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle ", angle)
```

```python
restaurant_types = ["biryani", "chinese", "pizza", "burger", "cafe"]
vectorized_upper = np.vectorize(str.upper)
print("Vectorized Upper ", vectorized_upper(restaurant_types))
```

```python
# Broadcast
monthly_avg = sales_data[:,1:] /12
print(monthly_avg)
```
